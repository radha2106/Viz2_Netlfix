### **📅 Calendar process steps 📅**

##### This code creates a csv file with dates along with its dependecies like year, month,day,ect. It asks you enter the desired dates (start and end dates)
---

```python
%%time

#--Time code processing
import time
start_time = time.perf_counter()

#--Step 1: Libraries
import pandas as pd
import numpy as np
import os
import warnings
warnings.filterwarnings('ignore')

#--Step 2: Inputs parameters (start date, end date and index start)
start_date = '2020/01/01' #input("What is the first date? (YYYY-MM-DD) ")
end_date = '2025/12/31' #input("What is the last date? (YYYY-MM-DD) ")
calendar_index = 1 #int(input("What was the calendar index?: "))

#--Step 3: Dates parameters range list
dates = pd.date_range(start=start_date, end=end_date)
idx = list(range(calendar_index, calendar_index + len(dates)))

#--Step 4: Add months in spanish (only if needed)
es_month_initials = {
    1: 'E', 2: 'F', 3: 'M', 4: 'A', 5: 'M', 6: 'J',
    7: 'J', 8: 'A', 9: 'S', 10: 'O', 11: 'N', 12: 'D'
}

#--Step 5: DataFrame Columns
years = dates.year
months = dates.month
months_names = dates.month_name().str[:3].str.strip()
days = dates.day
days_week = dates.dayofweek + 1
days_name = dates.day_name().str[:3].str.strip()
months_ends = dates + pd.offsets.MonthEnd(0)
quarters = dates.quarter
week_start = dates.to_period('W-SAT').start_time
week_end = dates.to_period('W-SAT').end_time.date
weeks =  np.where(dates.dayofweek + 1 > 5, 'Weekend','Weekday')
week_year = dates.isocalendar().week
month_year =  dates.month_name().str[:3].str.strip() + "-" + dates.year.astype('str').str[-2:].str.strip()
ini_month = dates.month.map(es_month_initials)
max_year_val = np.where(dates.year.min() == dates.year.max() , dates.year.min(), dates.year.max())
max_month_val = np.where(dates.month.min() == dates.month.max(), dates.month.min(), dates.month.max())
max_week_val = week_start.max()

#--Step 6: Calendar Dataframe and its columns
calendar = pd.DataFrame({
    'idx': idx,
    'dates': dates,
    'years': years,
    'months': months,
    'months_name': months_names,
    'days': days,
    'days_week': days_week,
    'days_name': days_name,
    'months_ends': months_ends,
    'quarters': quarters,
    'week_start': week_start,
    'week_end': week_end,
    'weeks': weeks,
    'week_year': week_year,
    'month_year': month_year,
    'ini_month': ini_month,
    'year_offset': years - max_year_val,
    'month_offset': months - max_month_val,
    'week_offset': (week_start - max_week_val).days // 7
}).reset_index(drop=True)
# --Fix any incorrect column type
calendar = calendar.astype({
    'week_start': 'datetime64[ns]',
    'week_end': 'datetime64[ns]',
    'week_year': 'int8',
    'week_offset': 'int8'
})

#--Optional Step: Getting the minimun and maximun years of dataframe to create name file
min_year = calendar['years'].min()
max_year = calendar['years'].max()

#--Step 7: File Name and Output Folder Path
filename = "calendar.csv" #f"calendar.csv" if min_year == max_year else f"calendar_{min_year}_to_{max_year}.csv"
output_folder = (r"--folderpath") #<< Always check the folder path
os.makedirs(output_folder, exist_ok=True)
file_path = os.path.join(output_folder, filename)

#--Step 8: Dataframe to CSV file
calendar.to_csv(file_path, index=False, encoding='utf-8')
print(f"File saved successfully as: {filename} with {len(calendar)} rows and {len(calendar.columns):.0f} columns; go and check your folder.")

end_time = time.perf_counter()
total_time = end_time - start_time
print(f"Code took {total_time:.2f} seconds to complete.")
```

### 🔁Renewal creation list🔁
##### This code creates a csv file that return the renewal dates and plan from the user file entered.

```python
%%time

#--Time code processing
import time
start_time = time.perf_counter()

#--Step 1: Libraries
import pandas as pd
import numpy as np
import os
import warnings
warnings.filterwarnings('ignore')

#--Random seed
rng = np.random.default_rng(seed=200)

#--Step 2: Get files path to be used, these files help you to get the information needed to create the output file
path_user = r"--folderpath"
path_plan = r"--folderpath"

#--Step 3: From step 2 convert the file to dataframes and get only columns needed from files
users = pd.read_csv(path_user, usecols=[0,1,3])
plans = pd.read_csv(path_plan,usecols=[0,3])

#--Step 4: Lower memory usage data type of the dataframes
plans = plans.astype({
    'idx':'int8',
    'daysplan':'int16'})
users = users.astype({
    'idx':'int32',
    'created_on':'int16',
    'plan_id':'int8'
})

#--Step 5: Merge both User and Plan into a new dataFrame
user_plan_df = pd.merge(users,plans, left_on='plan_id', right_on='idx',how='inner')

#--Step 6: Get needed columns from the merge dataframe
user_plan_df = user_plan_df[['idx_x','created_on','idx_y','daysplan']]

#--Step 7: Rename columns for better readability
user_plan_df = user_plan_df.rename(columns={'idx_x':'user_id','idx_y':'plan_id','created_on':'started_on'})

#--Step 8: Add 2 new columns, both need to be random integer value
user_plan_df = user_plan_df.assign(
    rnw_status = rng.integers(0,2, size=len(user_plan_df)).astype('int8'), # random value of 0-1 (inactive/active)
    rnd_nbr = lambda x: np.where(
        x['rnw_status'] == 1, rng.integers(1,4, size=len(x)),0).astype('int8') # random value of 1-3 for quantity of renewal
)

#--Step 9: Filter non-renewing rows from the original dataFrame, and add a new column to get last date of renewal
no_renewal = user_plan_df[(user_plan_df['rnw_status'] == 0) | (user_plan_df['rnd_nbr'] == 0)].copy()
no_renewal['last_date'] = no_renewal['started_on'] + no_renewal['daysplan'] #last date of plan
no_renewal = no_renewal[['user_id','plan_id', 'started_on','last_date', 'rnw_status']] #columns needed

#--Step 10: Filter for rows that require renewal expansion.
renewal = user_plan_df[(user_plan_df['rnw_status'] == 1) & (user_plan_df['rnd_nbr'] > 0)].copy()

#---Step 11: Create a list of renewal numbers for each row and explode the DataFrame.
renewal['rnw_nbr'] = renewal['rnd_nbr'].apply(lambda x: np.arange(1, x + 1))
renewal = renewal.explode('rnw_nbr')

#--Step 12: Calculate the new 'start_on' and 'end_on' dates for each renewal.
renewal['end_on'] = renewal['started_on'] + (renewal['rnw_nbr'] * renewal['daysplan']).astype('int16')
renewal['start_on'] = renewal['end_on'] - renewal['daysplan'].astype('int16')

#--Step 13: Update the 'rnw_status' for the last renewal.
renewal['rnw_status'] = np.where(renewal.groupby('user_id')['rnw_nbr'].transform('max') == renewal['rnw_nbr'], 0, 1).astype('int8')

#--Step 14: Select and reorder columns for the final result.
renewal = renewal[['user_id','plan_id', 'start_on','end_on', 'rnw_status']]

#--Step 15: Rename columns for better merges on both dataframes
no_renewal = no_renewal.rename(columns={'started_on':'plan_starts','last_date':'plan_ends'})
renewal = renewal.rename(columns={'start_on':'plan_starts','end_on':'plan_ends'})

#--Step 18: Combine both dataframes.
renewal_df = pd.concat([renewal, no_renewal], ignore_index=True).sort_values(by=['user_id']).reset_index(drop=True)

#--Step 19: Output file name and file output folder path
filename = 'renewal.csv'
output_folder = r"--folderpath" # Always check this folder path
os.makedirs(output_folder, exist_ok=True)
file_renewal = os.path.join(output_folder, filename)

#--Step 20: CSV creation
renewal_df.to_csv(file_renewal, index=False, encoding='utf-8-sig')
print(f"File saved successfully as: {filename} with {len(renewal_df)} rows go and check your folder.")

end_time = time.perf_counter()
total_time = end_time - start_time
print(f"Code took {total_time:.2f} seconds to complete.")
```

### **👁️Views Creation process👁️**
##### This code creates a csv file that return the views of each user, what program saw and its rating
---

```python
%%time

#--Time code processing
import time
start_time = time.perf_counter()

#--Step 1: Libraries needed
import pandas as pd
import numpy as np
import os
import warnings
warnings.filterwarnings('ignore')

#--Step 2: Enter inputs for quantity of rows for output file
rng = np.random.default_rng(seed=200)
n_rows = rng.integers(100000,500000,endpoint=True)

#--Step 3: Files path to be used, these files help you to get the information needed to create the output file
path_devices = r"--folderpath"
path_renewal = r"--folderpath"
path_program = r"--folderpath"
path_rating = r"--folderpath"

#--Step 4: Convert the path to dataframes and get only columns needed
device = pd.read_csv(path_devices,usecols=[0])
renewal = pd.read_csv(path_renewal,usecols=[0,2,3])
programs = pd.read_csv(path_program,usecols=[0,1])
rating = pd.read_csv(path_rating,usecols=[0])

#--Step 5: List of Variables from Series Frames, Probabilities and Random Choices
    #>> Device list, probabilty and choice
list_device = list(device['idx'])
prob_device = np.random.dirichlet(np.ones(len(list_device)))
device_type = np.random.choice(list_device, size=n_rows, p=prob_device)
    #>> Programs list, probabilty and choice
list_prog = list(programs['idx'])
prob_prog = np.random.dirichlet(np.ones(len(list_prog)))
prog_view = np.random.choice(list_prog, size=n_rows, p=prob_prog)
    #>> Rating list, probabilty and choice
list_rate = list(rating['idx'])
prob_rate = np.random.dirichlet(np.ones(len(list_rate)))
rate_type = np.random.choice(list_rate, size=n_rows, p=prob_rate)
    #>> User list, probabilty and choice
list_user = list(renewal['user_id'])
prob_user = np.random.dirichlet(np.ones(len(list_user)))
users_id = np.random.choice(list_user, size=n_rows, p=prob_user)

#--Step 6: Lower Memory Usage
device = device.astype({'idx':'int8'})
rating = rating.astype({'idx':'int8'})
renewal = renewal.astype({
    'user_id':'int64',
    'plan_starts':'int16',
    'plan_ends':'int16'})
programs = programs.astype({
    'idx':'int32',
    'uploaded_on':'int16'})

#--Step 7: Group users with their first and last date of subscription/renewal date; and drop unneeded columns
renewal = renewal.assign(
    min_start = renewal.groupby(['user_id'])[['plan_starts']].transform(min),
    max_start = renewal.groupby(['user_id'])[['plan_ends']].transform(max)
)
renewal = (renewal.drop(columns=['plan_starts','plan_ends'],inplace=False)
    .drop_duplicates(inplace=False))

#--Step 10: Create the output dataframe using steps 5-9
view_df = pd.DataFrame({
    'user_id': users_id,
    'progr_id': prog_view,
    'dev_id': device_type,
    'rate_id': rate_type
})

#--Step 11: Merge new dataframe with user and program to get dates columns
view_df = (view_df
        .merge(renewal, on='user_id',how='inner')
        .merge(programs, left_on='progr_id',right_on='idx',how='inner')
        .drop(columns=['idx'],inplace=False)
    )

#--Step 12: Create the view date column using the min_start and max_end columns, also create a helper Boolean column
    #to help filter view that are not possible due to user cannot see a program that is not released yet
view_df = (view_df
        .assign(
        viewed_on = lambda x: rng.integers(x['min_start'],x['max_start'],endpoint=True),
        filt_col = lambda x: (x['uploaded_on']>x['viewed_on']))
        .query('filt_col == 0')
        .drop(columns=['min_start','max_start','uploaded_on','filt_col'],inplace=False)
    ).reset_index(drop=True)

#--Step 13: Lower Memory Usage of new dataframe
view_df = view_df.astype({
    'dev_id':'int8',
    'rate_id':'int8',
    'viewed_on':'int16'
})

#--Step 14: Output file name, please change names for new ones when creating more than one file to not override existing files
filename = "renewal.csv"
output_folder = r"--folderpath"
os.makedirs(output_folder, exist_ok=True)
file_views = os.path.join(output_folder, filename)

#--Step 14: CSV creation
view_df.to_csv(file_views, index=False, encoding='utf-8-sig')
print(f"File saved successfully as: {filename} with {len(view_df)} rows go and check your folder.")

end_time = time.perf_counter()
total_time = end_time - start_time
print(f"Code took {total_time:.2f} seconds to complete.")
```
